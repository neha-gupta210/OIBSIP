```python
import pandas as pd
import numpy as np

df = pd.read_csv("Retail Sales Dataset.csv")

df.head()
```

```python
df.info()
```

```python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())
```

```python
df.describe()
```

```python
numeric_cols = ["Age","Quantity","Price per Unit","Total Amount","Total Price"]

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(col, "-> Outliers:", len(outliers))
```

```python
Q1  = df["Quantity"].quantile(0.25)
Q3 = df["Quantity"].quantile(0.75)
IQR = Q3 - Q1

upper = Q3 + 1.5 * IQR

print("Q1:",Q1)
print("Q3:",Q3)
print("IQR",IQR)
print("Upper limit", upper)

outliers = df[df["Quantity"] > upper]

print("Number of Quantity outliers:", len(outliers))
```

```python
outliers = df[df["Quantity"] > upper]

outliers[["Order ID", "Quantity", "Price per Unit", "Total Amount"]]
```

```python
outliers[["Order ID", "Quantity", "Price per Unit", "Total Amount", "Total Price"]]
```

```python
df["Calculated Amount"] = df["Quantity"] * df["Price per Unit"]

df[df["Calculated Amount"] != df["Total Amount"]][
    ["Order ID", "Quantity", "Price per Unit", "Total Amount", "Calculated Amount"]
]
```

```python
df["Date"] = pd.to_datetime(df["Date"], dayfirst=True)
```

```python
df["Date"].dtype
```

```python
summary = pd.DataFrame({
    "Metric": ["Rows", "Columns", "Missing Values", "Duplicate Rows"],
    "Before": [1000, 16, 0, 0],
    "After": [df.shape[0], df.shape[1], df.isnull().sum().sum(), df.duplicated().sum()]
})

summary
```

```python
df.drop(columns=["Calculated Amount"], inplace=True)
```

```python
df.to_csv("Retail_Sales_Cleaned.csv", index=False)
```

## Observation

- Dataset contains 1000 rows and 16 columns.
- No missing values were found.
- No duplicate rows were found.
- Date column was converted from string to datetime format.
- IQR method detected some outliers in Quantity and Total Amount.
- Quantity and Total Amount were checked for consistency.
- Detected outliers were retained for further analysis.
- The cleaned dataset was saved as Retail_Sales_Cleaned.csv.
